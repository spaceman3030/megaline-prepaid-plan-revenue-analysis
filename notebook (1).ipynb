```python
# Fill all NaN values before calculation with 0, NaN values are most likley user who haven't used any data
customer_usage_by_plan['call_minutes'] = customer_usage_by_plan['call_minutes'].fillna(0)
customer_usage_by_plan['messages_per_month'] = customer_usage_by_plan['messages_per_month'].fillna(0)
customer_usage_by_plan['gb_used'] = customer_usage_by_plan['gb_used'].fillna(0)
```

```python
# Calculate the monthly revenue for each user 
customer_usage_by_plan['messages_per_month'] = customer_usage_by_plan['messages_per_month'].fillna(0) # replacing NaN values with 0 for stat calc
def user_monthly_rev(row):
    if row['plan'] == 'surf':
        base_fee = 20 
         # calculating minute overage charge for surf plan
        if row['call_minutes'] <= 500: 
            minute_overage = 0
        else:
            minute_overage = (row['call_minutes'] - 500) * 0.03
            #calculating message overage for surf plan 
        if row['messages_per_month'] <= 50:
            message_overage = 0 
        else: 
            message_overage = (row['messages_per_month'] - 50) * 0.03
        if row['gb_used'] <= 15: 
            data_overage = 0
        else:
            data_overage = math.ceil(row['gb_used'] - 15) * 10
    elif row['plan'] == 'ultimate':
        base_fee = 70
        #calculating minute overage for ultimate plan 
        if row['call_minutes'] <= 3000:
            minute_overage = 0
        else: 
            minute_overage = (row['call_minutes'] - 3000) * 0.01
        if row['messages_per_month'] <= 1000:
            message_overage = 0
        else: 
            message_overage = (row['messages_per_month'] - 1000) * 0.01
        if row['gb_used'] <= 30:
            data_overage = 0 
        else:
            data_overage = row['gb_used'] - 30
            data_overage = data_overage * 7
    return base_fee + minute_overage + message_overage + data_overage           
customer_usage_by_plan['monthly_revenue'] = customer_usage_by_plan.apply(user_monthly_rev, axis=1)

print("First 10 rows with monthly revenue:")
print(customer_usage_by_plan[['user_id', 'month', 'plan', 'call_minutes', 'messages_per_month', 'gb_used', 'monthly_revenue']].head(10))

print("\nSummary statistics for monthly revenue:")
```

```python
# Test the hypotheses
# H0-ultimate and surf have no significant difference in revenue, H1-ultimate and surf significaly differ
# using the st.ttest.ind() to compare 2 population means, alpha set at 5% for accurate significance
from scipy import stats as st
import numpy as np

# creating variable for surf data plan  
surf_revenue = customer_usage_by_plan[customer_usage_by_plan['plan'] == 'surf']['monthly_revenue']


# creating variable for ultimate data plan 
ultimate_revenue = customer_usage_by_plan[customer_usage_by_plan['plan'] == 'ultimate']['monthly_revenue']

# set satistical significance 
alpha = 0.05

results = st.ttest_ind(surf_revenue, ultimate_revenue, equal_var=False)

print('p-value:', results.pvalue)

if results.pvalue < alpha:
    print('We reject the null hypothesis ')
else:
    print('We cant reject the null hypothesis ')
    
# investigating missing values causing a problems with calculations

print("Missing values in surf_revenue:", surf_revenue.isna().sum())
print("Missing values in ultimate_revenue:", ultimate_revenue.isna().sum())
print("Total surf revenue entries:", len(surf_revenue))
print("Total ultimate revenue entries:", len(ultimate_revenue))

# checking min and max for surf and ultimate revenue
print("Surf revenue - min:", surf_revenue.min(), "max:", surf_revenue.max())
print("Ultimate revenue - min:", ultimate_revenue.min(), "max:", ultimate_revenue.max())
print("Any infinite values in surf?", np.isinf(surf_revenue).any())
print("Any infinite values in ultimate?", np.isinf(ultimate_revenue).any())

# checking for missing values in call_time_duration, messages_per_month, and mb_used 
print("Missing values in call_minutes:", customer_usage_by_plan['call_minutes'].isna().sum())
print("Missing values in messages_per_month:", customer_usage_by_plan['messages_per_month'].isna().sum()) 
print("Missing values in gb_used:", customer_usage_by_plan['gb_used'].isna().sum())

missing_revenue_rows = customer_usage_by_plan[customer_usage_by_plan['monthly_revenue'].isna()]
print("Rows with missing revenue:")
print(missing_revenue_rows[['user_id', 'month', 'plan', 'call_minutes', 'messages_per_month', 'gb_used', 'monthly_revenue']].head(30))

# missing values indicate that calls or data wasn't used, changing missing values to 0 for stat calculation
```
